# 2 - Regressão Linear


Neste exemplo, queremos prever a potência gerada pelo painel solar a partir da irradiação solar.

Modelo desejado:

$$\hat{P} = \theta_0 + \theta_1 I$$

em que:

- $\hat{P}$ é a potência prevista, em watts;
- $I$ é a irradiação solar, em W/m²;
- $\theta_0$ é o intercepto da reta;
- $\theta_1$ é o coeficiente angular da reta.

```python
# We do select the irradiance variable as the independent variable (X) and the power variable as the dependent variable (y).

X = df[['Irradiação (W/m²)']]

# We can use other options such as df.iloc[:, 2] to select the third column such as varible X
# X = df.loc[:, 'Irradicação(W/m²)']  # This will select the column by name

# Now we select the dependent variable (y) as the power variable

Y = df[['P (W)']]

#Outras alternativas
#Y = df.iloc[:,3]  # Seleciona a quarta coluna (índice 3) como variável Y
#Y = df.loc[:, 'P (W)']   Seleciona a coluna de Potência (W) pelo nome
```
